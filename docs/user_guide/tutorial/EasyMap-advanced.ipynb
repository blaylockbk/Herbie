# 🗺️ EasyMap (more)

This companion notebook covers two useful features beyond the basics: an inset globe that locates the map extent, and NASA Global Imagery Browse Services (GIBS) imagery. For map creation, subplot layouts, data plotting, themes, and common layers, start with [EasyMap.ipynb](EasyMap.ipynb).


```python
from datetime import date

import matplotlib.pyplot as plt

from herbie.toolbox import EasyMap, ccrs, pc
```

## Add an inset globe

`INSET_GLOBE()` adds a small orthographic globe showing where the current map extent falls on Earth. Set the main map extent first. `x` and `y` position the inset in axes-relative coordinates, `size` controls its width and height, and `kind="area"` highlights the map footprint (`kind="point"` marks its center instead).


```python
fig, ax = plt.subplots(figsize=(9, 6), subplot_kw={"projection": ccrs.PlateCarree()})
map_ = EasyMap(ax=ax, theme="muted").OCEAN().LAND().STATES()
ax.set_extent([-130, -60, 20, 55], crs=pc)
map_.INSET_GLOBE(x=0.88, y=0.82, size=0.28, kind="area")
ax.set_title("CONUS extent with inset globe")
plt.show()
```

Use `kind="point"` when the center location is more useful than showing the full footprint. The optional `theme` and `facecolor` arguments customize the inset styling.


```python
fig, ax = plt.subplots(figsize=(9, 6), subplot_kw={"projection": ccrs.Mercator()})
map_ = EasyMap(ax=ax).OCEAN().LAND().BORDERS()
ax.set_extent([-15, 35, 30, 65], crs=pc)
map_.INSET_GLOBE(x=0.88, y=0.82, size=0.28, kind="point", facecolor="red")
ax.set_title("Europe extent with center-point inset")
plt.show()
```

## Add NASA Worldview imagery

`NASA()` retrieves an image from NASA GIBS and adds it to the map. This example uses the default VIIRS NOAA-20 true-color layer. It requires an internet connection and `scikit-image`; imagery availability depends on the selected date and layer. See the [NASA GIBS layer catalog](https://nasa-gibs.github.io/gibs-api-docs/available-visualizations/) for other options.


```python
fig, ax = plt.subplots(figsize=(9, 6), subplot_kw={"projection": ccrs.PlateCarree()})
map_ = EasyMap(ax=ax)
ax.set_extent([-130, -60, 20, 55], crs=pc)
map_.NASA(date(2022, 12, 10), pixels=1000).STATES().BORDERS()
ax.set_title("NASA GIBS true-color imagery: 2022-12-10")
plt.show()
```

The `date` argument may be a `datetime.date` or `datetime.datetime`. Increase `pixels` to request a larger image; larger requests may take longer. If the date is unavailable for the selected layer, choose a date supported by that imagery product.
