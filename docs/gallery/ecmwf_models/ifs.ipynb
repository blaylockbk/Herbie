# IFS

More examples getting data from ECMWF's IFS.

```python
from herbie import Herbie
```

By default, the IFS will retrieve the 0.1 degree operational **surface level** product, if the requested date is after October 20, 2026

```python
H = Herbie("2026-10-26", model="ifs")
print(H.grib)
print(f"{len(H.inventory()):,} fields")
```

The 0.1 degree *pressure level* data may also be requested.

```python
H = Herbie("2026-10-26", model="ifs", levtype="pl")
print(H.grib)
print(f"{len(H.inventory()):,} fields")
```

Otherwise, the 0.25 degree `oper` product is retrieved (both surface and pressure data are contained in the same file).

```python
H = Herbie("2026-10-1", model="ifs")
print(H.grib)
print(f"{len(H.inventory()):,} fields")
```
